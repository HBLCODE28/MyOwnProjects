##Install dependencies


```python
import subprocess
import sys

def run(cmd, label):
      print(f"[setup] {label}...")
      subprocess.run(cmd, shell=True, check=True)
      print(f"[setup] Done: {label}")

if IN_COLAB:
      print("[setup] Colab environment detected. Starting dependency setup.")

      run("pip install -r requirements.txt", "Install Python requirements")
      run("apt-get update -y", "Update apt packages")
      run("apt-get install -y ffmpeg", "Install ffmpeg")
      run("pip uninstall -y jax jaxlib || true", "Remove conflicting jax packages")
      run("pip install 'ml-dtypes>=0.3.1,<0.5.0'", "Install TensorFlow-compatible ml-dtypes")
      run("apt-get -qq update", "Quiet apt update")
      run("apt-get -qq install -y nodejs ffmpeg", "Install nodejs and ffmpeg")
      run("pip install -U yt-dlp", "Install yt-dlp")
      run("node -v", "Check node version")
      run("pip install flask-ngrok", "Install flask-ngrok")
      run('pip install "numpy<2" -U', "Pin numpy below version 2")
      run("pip install flask flask_sqlalchemy flask_login pyngrok", "Install Flask packages")
      run("pip install jupyter nbconvert notebook", "Install Jupyter tools")

      print("[setup] Colab dependency setup completed.")
```
